```python
import numpy as np
import random
import sys
import glob
import plotly.io as pio

pio.renderers.default = "notebook"
np.random.seed(0)
random.seed(0)
for path in glob.glob("../**/", recursive=True):
    sys.path.append(path)
```

When the Function is Convex and Differentiable, Finding their maximum / minimum is a easy task

```python
from src.algorithms import GradientDescent
from src.functions import Sphere
from src.utils import plots_3d

# Define the objective function, optimizer, and initial point
f = Sphere()
optimizer = GradientDescent(learning_rate=0.1)
x0 = np.array([75.0, 60.0])

# Run the optimization
result = optimizer.optimize(f, initial_guess=x0, stop_fitness=1e-6)

# Print the results
fig = plots_3d.plot_surface_and_path(f, result["x_history"], title="Gradient Descent on Sphere Function")
# Save as HTML (no extra dependencies required)
fig.show()
```

but what if the function has multiple minima?

```python
from src.algorithms import GradientDescent
from src.functions import Himmelblau
from src.utils import plots_3d

# Define the objective function, optimizer, and initial point
f = Himmelblau()
optimizer = GradientDescent(learning_rate=0.001)
x0 = np.array([4.0, 4.0])

# Run the optimization
result = optimizer.optimize(f, initial_guess=x0, stop_fitness=1e-6, minimize=True)

# Print the results
fig = plots_3d.plot_surface_and_path(f, result["x_history"], title="Gradient Descent on Himmelblau Function")
fig.show()
```
