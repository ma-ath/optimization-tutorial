```python
import numpy as np
import random
import sys
from pathlib import Path
from matplotlib import rc

np.random.seed(0)
random.seed(0)
rc('animation', html='jshtml')
sys.path.append(str(Path().resolve().parent))
```

When the Function is Convex and Differentiable, Finding their maximum / minimum is a easy task

```python
from src.algorithms import GradientDescent
from src.functions import Sphere
from src.utils import plots_3d
from IPython.display import HTML

# Define the objective function, optimizer, and initial point
f = Sphere()
optimizer = GradientDescent(learning_rate=0.1)
x0 = np.array([75.0, 60.0])

# Run the optimization
result = optimizer.optimize(f, initial_population=x0, stop_fitness=None, budget=100)

# Print the results
anim = plots_3d.animate_path_on_surface(f, result["x_history"], title="Gradient Descent on Sphere Function")
HTML(anim.to_jshtml())
```

but what if the function has multiple minima?

```python
from src.algorithms import GradientDescent
from src.functions import Himmelblau
from src.utils import plots_3d
from IPython.display import HTML

# Define the objective function, optimizer, and initial point
f = Himmelblau()
optimizer = GradientDescent(learning_rate=1e-3)
# x0 = np.array([[4.0, 4.0], [-4.0, -4.0], [4.0, -4.0], [-4.0, 4.0]])
x0 = np.random.uniform(-5, 5, (100, 2))

# Run the optimization
result = optimizer.optimize(f, initial_population=x0, stop_fitness=None, budget=300, minimize=True)

# Print the results
anim = plots_3d.animate_path_on_surface(f, result["x_history"], title="Gradient Descent on Himmelblau Function")
HTML(anim.to_jshtml())
```

```python
from src.algorithms import GaussianEvolutionStrategy
from src.functions import Himmelblau
from src.utils import plots_3d

# Define the objective function, optimizer, and initial point
f = Himmelblau()
optimizer = GaussianEvolutionStrategy(mean0=np.array([0.0, 0.0]), sigma0=5.0)

# Run the optimization
result = optimizer.optimize(f, population_size=10, function_dimension=2, budget=1000, minimize=True, stop_fitness=1e-6)

# Plot the results
anim = plots_3d.animate_population_on_surface(f, result["x_history"], title=f"{optimizer.name} on {f.name} Function")
HTML(anim.to_jshtml())
```

```python
from src.algorithms import DifferentialEvolution
from src.functions import Ackley, Beale, Griewank, Himmelblau, Levy, Rastrigin, Rosenbrock, Schwefel, Sphere
from src.utils import plots_3d
from IPython.display import HTML

functions = [Ackley(), Beale(), Griewank(), Himmelblau(), Levy(), Rastrigin(), Rosenbrock(), Schwefel(), Sphere()]
x_history = []

for f in functions:
    optimizer = DifferentialEvolution()
    result = optimizer.optimize(f, population_size=20, function_dimension=2, budget=500)
    x_history.append(result["x_history"])

anim = plots_3d.animate_population_on_grid_of_surfaces(functions, x_history, resolution=200)
HTML(anim.to_jshtml())
```
