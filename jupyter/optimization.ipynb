# Optimization

```python
import numpy as np
import random
import sys
from pathlib import Path
from matplotlib import rc

np.random.seed(0)
random.seed(0)
rc('animation', html='jshtml')
sys.path.append(str(Path().resolve().parent))
```

When the Function is Convex and Differentiable, Finding their maximum / minimum is a easy task

```python
from src.algorithms import GradientDescent
from src.functions import Sphere
from src.utils import plots_3d

# Define the objective function, optimizer, and initial point
f = Sphere()
optimizer = GradientDescent(learning_rate=0.1)
x0 = np.array([75.0, 60.0])

# Run the optimization
result = optimizer.optimize(f, initial_guess=x0, stop_fitness=1e-6)

# Print the results
plots_3d.animate_path_on_surface(f, result["x_history"], title="Gradient Descent on Sphere Function")
```

but what if the function has multiple minima?

```python
from src.algorithms import DifferentialEvolution
from src.functions import Ackley
from src.utils import plots_3d

# Define the objective function, optimizer, and initial point
f = Ackley()
optimizer = DifferentialEvolution()

# Run the optimization
result = optimizer.optimize(f, population_size=40, function_dimension=2, budget=1000, minimize=True)

# Plot the results
plots_3d.animate_population_on_surface(f, result["x_history"], title="Differential Evolution on Ackley Function")
```
